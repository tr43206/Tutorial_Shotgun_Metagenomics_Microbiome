```python
fig, ax = plt.subplots(figsize=(10,4))

# Stacked barplot (막대 간격 최소화: width=1)
df.plot(kind='bar', stacked=True, width=1.0, linewidth=0.5,
        edgecolor='darkgrey',
        ax=ax, colormap=GutLungAxis)

ax.set_xticks([])
ax.tick_params(axis='x', which='both', bottom=False, top=False)

ax.yaxis.grid(True, linestyle='dotted', linewidth=0.5, color='gray', zorder=0)
ax.set_axisbelow(True)

ax.set_xlabel('15 samples', fontsize=9)
ax.set_ylabel('Relative Abundance', fontsize=10)
ax.set_title('EC 4.4.1.21: S-ribosylhomocysteine lyase', fontsize=10.5)

# vlines (샘플 개수 및 범위 확인 후 위치 정확히 조정)
ymin, ymax = ax.get_ylim()
for x in [5.5]:  #Edit
    ax.vlines(x=x, ymin=ymin, ymax=ymax, linestyles='-', linewidth=0.5, colors='k')

legend = plt.legend(bbox_to_anchor=(1, 1.08), fontsize=9, ncol=2, loc='upper left', frameon=False)
legend.set_title('Stratified contributions:', prop={'size': 9})
legend._legend_box.align = "left"

ax2 = ax.inset_axes([0, -0.2, 1.05, 0.1])  # [left, bottom, width, height]

n_group1 = 6  #Edit
n_group2 = 9  #Edit
group_sizes = {'Control': n_group1, 'VNAM': n_group2}
group_colors = {'VNAM': '#87CEEB',
                #'FOS_BP': '#9ACD32',
                'Control': '#FFC0CB'}
start = 0
for group in ['Control', 'VNAM']:
    width = group_sizes[group]
    ax2.barh(0, width, left=start, color=group_colors[group], edgecolor='k', linewidth=0.5)
    ax2.text(start + width/2, -1, group.replace('_', ' '), ha='center', va='center', fontsize=9)
    start += width

ax2.axis('off')

# --- 박스 상하좌우 딱 맞게 조정 ---
ax.margins(0)  # Barplot 여백 제거
ax.set_xlim(-0.5, len(df)-0.5)  # X축 한 칸 여유만
ax.set_ylim(0, ymax)  # Y축 최소/최대값

plt.tight_layout()
plt.savefig('humann_barplot_ec-4.4.1.21.png', dpi=500, bbox_inches='tight')
```
