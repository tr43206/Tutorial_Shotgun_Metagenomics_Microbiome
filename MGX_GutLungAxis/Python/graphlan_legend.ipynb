```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from matplotlib.lines import Line2D

plt.rcParams['font.family'] = 'Arial'

import warnings
warnings.filterwarnings("ignore")
```

```python
group_names = ["VNAM", "Control"]  ##Edit: 그룹명 순서 반대로 넣어야 함
```

```python
# Graphlan annotation background color (phylum background sectors)
phylum_bg_colors = {"Pseudomonadota": "#B0FFC0",
                    "Actinomycetota": "#FFB0B0",
                    "Bacteroidota": "#B0E9FD",
                    "Firmicutes": "#FCDBA2",
                    "Other phylum": "#BEBEBE",
                    }

# Ring colors in order
ring_colors = ["#5ebcc2", "#829d45", "#d175ae", "cyan", "magenta", "yellow", "black", "white"]

# 실제 그림에 맞게 그룹 수만큼만 사용
group_colors = {name: ring_colors[i] for i, name in enumerate(group_names)}
```

```python
fig, ax = plt.subplots(figsize=(1.9, 2.4))
ax.axis("off")

# Phylum background legend
phylum_order = ["Firmicutes", "Actinomycetota", "Bacteroidota", "Pseudomonadota", "Other phylum"]
phylum_bg_handles = [Patch(facecolor=phylum_bg_colors[name], edgecolor='black', label=name)
                     for name in phylum_order]

# Group ring legend
group_order = ["Control", "VNAM"]
group_handles = [Patch(facecolor=group_colors[name], edgecolor='black', linewidth=0.5, label=name)
                 for name in group_order]

# Legend 1: Phylum (background)
leg1 = ax.legend(handles=phylum_bg_handles, title="Phylum ", loc="upper left", bbox_to_anchor=(-0.08, 1.10), frameon=False, fontsize=11, title_fontsize=12)
#leg1._legend_box.align = "left"
ax.add_artist(leg1)

# Legend 2: Groups
leg2 = ax.legend(handles=group_handles, title="Group", loc="upper left", bbox_to_anchor=(-0.08, 0.33), frameon=False, fontsize=11, title_fontsize=12)
#leg2._legend_box.align = "left"
ax.add_artist(leg2)

plt.tight_layout()
plt.savefig("C:/Users/tr432/Downloads/graphlan_legend.png", dpi=500, bbox_inches="tight")
```
